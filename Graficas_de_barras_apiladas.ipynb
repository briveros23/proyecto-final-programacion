```python
import matplotlib.pyplot as plt
import pandas as pd
from collections import Counter
import csv

def graf_apl(var_1='Weekly_study_hours',var_2='GRADE',titulo=f'Proporción de variable 1, respecto a variable 2',tx='Proporción',ty='Variable 2',tl='clasificación',clm_par='GRADE',parametro=None,rev=True):
  df = pd.read_csv('StudentsPerformance_with_headers.csv')
  variables = df.columns[1:]
  mapeo = {}
  with open('/content/var_codes_meanings_student_performance.csv', 'r', errors = 'ignore') as csvfile:
    lector_csv = csv.reader(csvfile)
    var_codes = list(lector_csv)
  for var in variables:
      mapeo[var] = {}
      for row in var_codes:
        if row[0] == var:
          mapeo[var][int(row[1])] = row[2]
  estudio=list(df[clm_par])
  cat=list(df[var_1])
  clf=list(df[var_2])
  contador_cat = Counter(cat)
  contador_cat = dict(sorted(contador_cat.items()))
  nhs={}
  for i in range(0,len(cat)):
            if parametro is None or estudio[i] == parametro:
              if cat[i] not in nhs:
                nhs[cat[i]] = [clf[i]]
              elif cat[i] in nhs:
                  nhs[cat[i]].append(clf[i])
                  nhs[cat[i]] = sorted(nhs[cat[i]])
  nhs = dict(sorted(nhs.items()))
  f=0
  frac_clf=[]
  k = "variable temporal"
  for j in list(nhs.values()):
    for d in range(sorted(list(Counter(clf).keys()))[0],sorted(list(Counter(clf).keys()))[-1]+1):
      if d not in j:
        if d!=sorted(list(Counter(clf).keys()))[0]:
          frac_clf[f].insert(d,0)
        else:
          frac_clf.append([sorted(list(Counter(clf).keys()))[0]])
      else:
        if d!=sorted(list(Counter(clf).keys()))[0]:
          if sorted(list(Counter(clf).keys()))[0] not in frac_clf[f]:
            if list(Counter(j).values()) in frac_clf:
             continue
            else:
              frac_clf.append(list(Counter(j).values()))
              k = frac_clf.append(list(Counter(j).values()))
          else:
            if k in frac_clf:
              continue
            else:
              if frac_clf[f]==[sorted(list(Counter(clf).keys()))[0]]:
                k=list(Counter(j).values())
                frac_clf[f]=[sorted(list(Counter(clf).keys()))[0]]+list(Counter(j).values())
        else:
          frac_clf.append(list(Counter(j).values()))
          k = frac_clf
    f+=1
  fracc_clf = []
  ñ=0
  for l in frac_clf:
    fracc_clf.append([])
    for p in l:
      fracc_clf[ñ].append(float(f"{p/int(list(contador_cat.values())[ñ]):.3f}"))

    ñ+=1

  proporciones=[]
  ñ=0
  print(fracc_clf)
  for q in range(0,len(fracc_clf[0])):
    proporciones.append([])
    for w in range(0,len(fracc_clf)):
        proporciones[q].append((fracc_clf[w][q]))

  categorias = []
  for i,j in mapeo[var_1].items():
    categorias.append(j)

  clasif={}
  ñ=0
  for i,j in mapeo[var_2].items():
    clasif[j]=proporciones[ñ]
    ñ+=1

  hulu = pd.DataFrame(clasif, index=categorias)

  hulu.plot(kind='barh', stacked=True, colormap='viridis')

  plt.xlabel(tx)
  plt.ylabel(ty)
  plt.title(titulo)
  plt.legend(title=tl, bbox_to_anchor=(1, 1), loc='upper left',reverse=rev)
  plt.show()

graf_apl('GRADE','Listening in classes','Proporción de frecuencia de los estudiantes para escuchar en clase, con respecto a las calificaciones','Proporción','Calificaciones','Frecuencia')
```
