```python
import matplotlib.pyplot as plt
import csv
import math

with open('/content/updated_students_performance_dataset.csv', 'r') as csvfile:
  csv_reader = csv.reader(csvfile)
  data = list(csv_reader)
```

```python
trabajo_adicional = data[0].index('Additional_work')
salario_mensual = data[0].index('Total_salary_if_available')
transporte = data[0].index('Transportation to the university')
tipo_escuela = data[0].index('Graduated_high_school_type')
actividad_extra = data[0].index('Regular artistic or sports activity')

num_estudiantes = range(1, len(data))

confort_alto = 0
confort_medio = 0
confort_bajo = 0

for estudiante in num_estudiantes:
  nivel_confort = 0
  if data[estudiante][trabajo_adicional] == 'No':
    nivel_confort += 10*0.25
  if data[estudiante][salario_mensual] == 'USD 135-200':
    nivel_confort += 2*0.3
  elif data[estudiante][salario_mensual] == 'USD 201-270':
    nivel_confort += 4*0.3
  elif data[estudiante][salario_mensual] == 'USD 271-340':
    nivel_confort += 6*0.3
  elif data[estudiante][salario_mensual] == 'USD 341-410':
    nivel_confort += 8*0.3
  elif data[estudiante][salario_mensual] == 'above 410':
    nivel_confort += 10*0.3
  if data[estudiante][transporte] == 'Bus':
    nivel_confort += 4*0.2
  elif data[estudiante][transporte] == 'Bicycle':
    nivel_confort += 4*0.2
  elif data[estudiante][transporte] == 'Private car/taxi':
    nivel_confort += 10*0.2
  if data[estudiante][tipo_escuela] == 'private':
    nivel_confort += 10*0.125
  if data[estudiante][tipo_escuela] == 'other':
    nivel_confort += 5*0.125
  if data[estudiante][actividad_extra] == 'Yes':
    nivel_confort += 10*0.125
  if nivel_confort >= 0 and nivel_confort <= 3.9:
    confort_bajo += 1
  elif nivel_confort >= 4 and nivel_confort <= 6.9:
    confort_medio += 1
  else:
    confort_alto += 1
```

```python
AA = 0
BA = 0
BB = 0
CB = 0
CC = 0
DC = 0
DD = 0
FALLO = 0
CC = 0

calificaciones = data[0].index('GRADE')

for row in data:
  if row[calificaciones] == 'AA':
    AA += 1
  elif row[calificaciones] == 'BA':
    BA += 1
  elif row[calificaciones] == 'BB':
    BB += 1
  elif row[calificaciones] == 'CB':
    CB += 1
  elif row[calificaciones] == 'CC':
    CC += 1
  elif row[calificaciones] == 'DC':
    DC += 1
  elif row[calificaciones] == 'DD':
    DD += 1
  else:
    FALLO += 1

calificaciones = ['AA', 'BA', 'BB', 'CB', 'CC', 'DC', 'DD', 'FALLO']
valores = [AA, BA, BB, CB, CC, DC, DD, FALLO]

colores = [
    "#FF6347",
    "#FFA07A",
    "#FFD700",
    "#98FB98",
    "#87CEFA",
    "#DDA0DD",
    "#FFB6C1",
    '#40E0D0'
]
```

```python
asistencia = data[0].index('Attendance_to_classes')
escucha_activa = data[0].index('Listening in classes')
apuntes = data[0].index('Taking notes in classes')
dialogos_constructivos = data[0].index('Discussion improves my interest and success in the course')

num_estudiantes = range(1, len(data))

interes_alto = 0
interes_medio = 0
interes_bajo = 0

for estudiante in num_estudiantes:
  interes = 0
  if data[estudiante][asistencia] == 'Always':
    interes += 10*0.3
  elif data[estudiante][asistencia] == 'Sometimes':
    interes += 5*0.3
  if data[estudiante][escucha_activa] == 'Always':
    interes += 10*0.3
  elif data[estudiante][escucha_activa] == 'Sometimes':
    interes += 5*0.3
  if data[estudiante][apuntes] == 'Always':
    interes += 10*0.1
  elif data[estudiante][apuntes] == 'Sometimes':
    interes += 5*0.1
  if data[estudiante][dialogos_constructivos] == 'Always':
    interes += 10*0.3
  elif data[estudiante][dialogos_constructivos] == 'Sometimes':
    interes += 5*0.3
  if interes >= 0 and interes <= 3.9:
    interes_bajo += 1
  elif interes >= 4 and interes <= 6.9:
    interes_medio += 1
  else:
    interes_alto += 1
```

```python
tiempo = data[0].index('Weekly_study_hours')
lectura_cientifica = data[0].index('Reading_frequency_science')
seminarios = data[0].index('Attendance_to_the_seminars')
preparacion_examen = data[0].index('Preparation to midterm exams 2')

num_estudiantes = range(1, len(data))

intensidad_alta = 0
intensidad_media = 0
intensidad_baja = 0

for estudiante in num_estudiantes:
  intensidad = 0
  if data[estudiante][tiempo] == 'Less than 5 hours':
    intensidad += 3*0.3
  elif data[estudiante][tiempo] == '6-10 hours':
    intensidad += 5.5*0.3
  elif data[estudiante][tiempo] == '11-20 hours':
    intensidad += 8*0.3
  elif data[estudiante][tiempo] == 'More than 20 hours':
    intensidad += 10*0.3
  if data[estudiante][lectura_cientifica] == 'Often':
    intensidad += 10*0.3
  elif data[estudiante][lectura_cientifica] == 'Sometimes':
    intensidad += 5*0.3
  if data[estudiante][seminarios] == 'Yes':
    intensidad += 10*0.2
  if data[estudiante][preparacion_examen] == 'Regularly during the semester':
    intensidad += 10*0.2
  elif data[estudiante][preparacion_examen] == 'Closest date to the exam':
    intensidad += 7*0.2
  if intensidad >= 0 and intensidad <= 3.9:
    intensidad_baja += 1
  elif intensidad >= 4 and intensidad <= 6.9:
    intensidad_media += 1
  else:
    intensidad_alta += 1
```

```python
num_hermanos = data[0].index('Number_of_sisters/brothers')
pareja = data[0].index('Do you have a partner')
estado_parental = data[0].index('Parental_status')
modo_de_estudio = data[0].index('Preparation to midterm exams 1')
trabajo_adicional = data[0].index('Additional_work')

num_estudiantes = range(1, len(data))

apoyo_social_alto = 0
apoyo_social_medio = 0
apoyo_social_bajo = 0

for estudiante in num_estudiantes:
  apoyo_social = 0
  if data[estudiante][num_hermanos] != '0':
    apoyo_social += 10*0.154
  if data[estudiante][pareja] == 'Yes':
    apoyo_social += 10*0.3
  if data[estudiante][estado_parental] == 'Married':
    apoyo_social += 10*0.1
  if data[estudiante][modo_de_estudio] == 'With friends':
    apoyo_social += 10*0.35
  if data[estudiante][trabajo_adicional] == 'No':
    apoyo_social += 10*0.1
  if apoyo_social >= 0 and apoyo_social <= 3.9:
    apoyo_social_bajo += 1
  elif apoyo_social >= 4 and apoyo_social <= 6.9:
    apoyo_social_medio += 1
  else:
    apoyo_social_alto += 1
```

```python
niveles = ['Bajo', 'Medio', 'Alto']

valores_1 = [confort_bajo, confort_medio, confort_alto]
valores_2 = [interes_bajo, interes_medio, interes_alto]
valores_3 = [intensidad_baja, intensidad_media, intensidad_alta]
valores_4 = [apoyo_social_bajo, apoyo_social_medio, apoyo_social_alto]

colors = [
    '#F6C1C1',
    '#FFB347',
    '#FF6F61',
]

fig, axs = plt.subplots(2, 2, figsize=(6.5, 6.5))

fig.suptitle('Caracterización Poblacional', fontsize = 16, y = 1)

axs[0, 0].pie(valores_1, labels=None, colors=colors, wedgeprops={'linewidth': 3, 'edgecolor': 'white'}, startangle = 90, autopct='%1.1f%%', pctdistance = 1.2)
axs[0, 0].add_artist(plt.Circle((0, 0), 0.5, color='white'))
axs[0,0].set_title('Confort')

axs[0, 1].pie(valores_2, labels=None, colors=colors, wedgeprops={'linewidth': 3, 'edgecolor': 'white'}, startangle = 45, autopct='%1.1f%%', pctdistance = 1.2)
axs[0, 1].add_artist(plt.Circle((0, 0), 0.5, color='white'))
axs[0,1].set_title('Interés en sus Cursos')

axs[1, 0].pie(valores_3, labels=None, colors=colors, wedgeprops={'linewidth': 3, 'edgecolor': 'white'}, startangle = 30, autopct='%1.1f%%', pctdistance = 1.2)
axs[1, 0].add_artist(plt.Circle((0, 0), 0.5, color='white'))
axs[1,0].set_title('Intensidad de Estudio')

axs[1, 1].pie(valores_4, labels=None, colors=colors, wedgeprops={'linewidth': 3, 'edgecolor': 'white'}, startangle = 70, autopct='%1.1f%%', pctdistance = 1.2)
axs[1, 1].add_artist(plt.Circle((0, 0), 0.5, color='white'))
axs[1,1].set_title('Apoyo Social')

fig.legend(niveles, loc='upper right', bbox_to_anchor=(1.1, 0.85), title="Nivel")

plt.savefig('caracterizacion_poblacional.png')

plt.tight_layout()
plt.show()
```
